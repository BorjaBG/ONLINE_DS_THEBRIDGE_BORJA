## Ejercicio 1
Somos una empresa que tiene varias tiendas de electrodomesticos. Necesitamos un programa para manejar las tiendas, ver las ventas que han tenido, cuántos empleados hay, etc... Para ello vamos a modelizar la tienda en una clase, que tendrá las siguientes características:
* Nombre clase: "Tienda"
* Atributos comunes:
    * Tipo: "Electrodomésticos"
    * Abierta: True
* Atributos propios de cada tienda:
    * Nombre - String
    * Dirección - String
    * Número de empleados - int
    * Ventas ultimos 3 meses - Lista de 3 numeros
* Método para calcular las ventas de todos los meses, que devuelva un numero con todas las ventas.
* Método que calcula la media de ventas de los ultimos meses, por empleado
* Método que devuelve en un string el nombre de la tienda, junto con su dirección.
* Método que obtiene las ventas del último mes.
* Método para dar una proyección de las ventas en caso de haber invertido X dinero en marketing en esa tienda. Siendo X el parámetro de entrada. Si X es menor de 1000, las ventas de los ultimos 3 meses hubiesen sido de (1.2 x ventas), si es mayor o igual de 1000, las ventas hubiesen sido de (1.5 x venta). El método reescribe el atributo de ventas en los últimos 3 meses, y además devuelve las ventas con el aumento.

Se pide:
1. Implementar la clase "Tienda"
2. Crear tres tiendas con datos inventados
3. Comprueba en al menos una de ellas, todo lo implementado en la clase tienda (sus atributos, media de ventas, ventas/empleado...)
3. Calcular las ventas del último mes de todas las tiendas. Para ello usa el bucle `for`
4. Imprime por pantalla los nombres de las tiendas cuya dirección lleve el string "Avenida"
5. Seguro que ya has hecho este apartado... Documenta la clase :)

```python
class Tienda:
    '''
    Clase que representa una tienda de electrodomésticos.

    Atributos de clase:
        tipo (str): Tipo de tienda, por defecto 'Electrodomésticos'.
        abierta (bool): Estado de la tienda, por defecto True.

    Atributos de instancia:
        nombre (str): Nombre de la tienda.
        direccion (str): Dirección física de la tienda.
        numero_empleados (int): Número de empleados que trabajan en la tienda.
        ventas_ultimos_3_meses (list): Lista con las ventas de los últimos 3 meses.
    '''
    tipo = 'Electrodomésticos'
    abierta = True

    def __init__(self, nombre, direccion, numero_empleados, ventas_ultimos_3_meses):
        '''
        Inicializa una nueva instancia de la clase Tienda.
        Args:
            nombre (str): Nombre de la tienda.
            direccion (str): Dirección física de la tienda.
            numero_empleados (int): Número de empleados que trabajan en la tienda.
            ventas_ultimos_3_meses (list): Lista con las ventas de los últimos 3 meses.
        '''
        self.nombre = nombre
        self.direccion = direccion
        self.numero_empleados = numero_empleados
        self.ventas_ultimos_3_meses = ventas_ultimos_3_meses

    def ventas_totales(self):
        '''
        Calcula y devuelve la suma total de las ventas de todos los meses.
        Returns:
            float: Suma total de las ventas.
        '''
        return sum(self.ventas_ultimos_3_meses)

    def ventas_por_empleado(self):
        '''
        Calcula la media de ventas de los últimos meses por cada empleado.
        Returns:
            float: Media de ventas por empleado.
        '''
        return self.ventas_totales() / self.numero_empleados

    def nombre_direccion(self):
        '''
        Devuelve una cadena con el nombre de la tienda y su dirección.
        Returns:
            str: Cadena con el nombre y dirección de la tienda.
        '''
        return f'Nombre: {self.nombre}, Dirección: {self.direccion}'

    def ventas_ultimo_mes(self):
        '''
        Obtiene las ventas correspondientes al último mes.
        Returns:
            float: Ventas del último mes.
        '''
        return self.ventas_ultimos_3_meses[-1]

    def proyeccion_ventas(self, inversion_marketing):
        '''
        Proyecta e incrementa las ventas según la inversión en marketing.

        Si la inversión es menor a 1000, las ventas aumentan un 20% (x1.2).
        Si es mayor o igual a 1000, aumentan un 50% (x1.5).
        Actualiza el atributo de ventas y lo devuelve.
        Args:
            inversion_marketing (float): Cantidad invertida en marketing.
        Returns:
            list: Lista actualizada de las ventas.
        '''
        if inversion_marketing < 1000:
            aumento = 1.2
        else:
            aumento = 1.5
        return [venta * aumento for venta in self.ventas_ultimos_3_meses]


tienda1 = Tienda('Tienda1', 'Calle 1', 10, [180, 431, 300])
tienda2 = Tienda('Tienda2', 'Avenida 2', 20, [200, 354, 410])
tienda3 = Tienda('Tienda3', 'Calle 3', 30, [275, 540, 721])

print(tienda1.ventas_totales())
print(tienda1.ventas_por_empleado())
print(tienda1.nombre_direccion())
print(tienda1.ventas_ultimo_mes())
print(tienda1.proyeccion_ventas(1000))

ventas_totales = 0
for tienda in [tienda1, tienda2, tienda3]:
    ventas_totales += tienda.ventas_totales()
print(f'Ventas totales de todas las tiendas: {ventas_totales}')

for tienda in [tienda1, tienda2, tienda3]:
    if 'Avenida' in tienda.direccion:
        print(tienda.nombre)
```

## Ejercicio 2
En este ejercicio vamos a implementar una clase *Perro* en Python. La clase tiene las siguientes características:
* Cosas que sabemos seguro que tiene un perro
    * Tiene 4 patas
    * 2 orejas
    * 2 ojos
    * Una velocidad de 0. Por defecto, el perro está parado
* Cuando se inicialice:
    * El perro será de una determinada raza
    * Por defecto tendrá pelo "Marrón", a no ser que se diga lo contrario.
    * Por defecto no tendrá dueño, a no ser que se diga lo contrario.
    
* Dispondrá también de un método llamado andar, que tiene un argumento de entrada (aumento_velocidad). Este valor se le sumará a la velocidad que ya llevaba el perro.
* Necesita otro método (parar), donde pondremos la velocidad a 0.
* Otro método llamado "ladrar", que tendrá un argumento de entrada, y la salida será el siguiente string: "GUAU!" + el argumento de entrada.


Se pide:
* Implementa la clase *Perro*
* Crea un objeto de tipo *Perro*, sin dueño
* Comprueba que están bien todos sus atributos
* Prueba que ande, y comprueba su velocidad
* Páralo
* Documenta la clase *Perro*

```python
class Perro:
  '''
  Clase que representa a un perro.

  Atributos de clase:
    patas (int): Número de patas del perro.
    orejas (int): Número de orejas del perro.
    ojos (int): Número de ojos del perro.

  Atributos de instancia:
    raza (str): Raza del perro.
    color_pelo (str): Color del pelo del perro.
    dueno (str): Nombre del dueño del perro.
    velocidad (int): Velocidad actual del perro.
  '''
  patas = 4
  orejas = 2
  ojos = 2

  def __init__(self, raza, color_pelo='Marrón', dueno=None, velocidad=0):
    '''
    Inicializa una nueva instancia de la clase Perro.

    Args:
      raza (str): Raza del perro.
      color_pelo (str): Color del pelo del perro.
      dueno (str): Nombre del dueño del perro.
      velocidad (int): Velocidad actual del perro.
    '''
    self.raza = raza
    self.color_pelo = color_pelo
    self.dueno = dueno
    self.velocidad = velocidad

  def andar(self, aumento_velocidad):
    '''
    Actualiza la velocidad del perro y devuelve la nueva velocidad.

    Args:
      aumento_velocidad (int): Cantidad a aumentar la velocidad.

    Returns:
      int: Nueva velocidad del perro.
    '''
    self.velocidad += aumento_velocidad
    return self.velocidad

  def parar(self):
    '''
    Establece la velocidad del perro en 0.
    '''
    self.velocidad = 0

  def ladrar(self, mensaje):
    '''
    Devuelve un mensaje con el perro ladrando.

    Args:
      mensaje (str): Mensaje a ladrar.

    Returns:
      str: Mensaje con el perro ladrando.
    '''
    return f'GUAU! {mensaje}'



perro1 = Perro('Pastor Alemán')

print(perro1.raza)
print(perro1.color_pelo)
print(perro1.dueno)
print(perro1.velocidad)

velocidad = perro1.andar(6)
print(velocidad)

perro1.parar()
```
