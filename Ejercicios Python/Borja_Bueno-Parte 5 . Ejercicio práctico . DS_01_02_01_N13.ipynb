## Ejercicio 12
Solicitar al usuario que ingrese su dirección email. Imprimir un mensaje indicando si la dirección es válida o no, valiéndose de una función para decidirlo. Una dirección se considerará válida si contiene el símbolo "@".

`````python
def validar_email(email):
    '''
    Esta función recibe un email y devuelve True si contiene el símbolo "@" y False si no lo contiene.
    Args:
        email (str): email a validar
    Returns:
        str: mensaje indicando si el email es válido o no
    '''
    if "@" in email:
        return "Email valido"
    else:
        return "Email invalido"

email = input("Introduce tu dirección de email: ")
resultado = validar_email(email)
print(resultado)
`````

## Ejercicio 13
Escribe una función que compruebe si un DNI es válido, devolviendo `True` en caso afirmativo.  
En primer lugar, el DNI debe tener entre 7 y 8 dígitos numéricos.  
Para saber si la letra del DNI es correcta puedes utilizar el siguiente fragmento de código:
>```Python
>DNI= 55555555
>palabra='TRWAGMYFPDXBNJZSQVHLCKE'
>print('La letra del DNI es: ', palabra[DNI%23])
>````


`````python
def validar_dni(dni):
    '''
    Esta función recibe un DNI y devuelve True si es válido y False si no lo es.
    Args:
        dni (str): DNI a validar
    Returns:
        str: mensaje indicando si el DNI es válido o no
    '''
    palabra = 'TRWAGMYFPDXBNJZSQVHLCKE'
    dni = dni.strip().upper()

    numeros = dni[:-1]
    letra = dni[-1:]

    if len(numeros) < 7 or len(numeros) > 8:
        return "DNI inválido"

    numero_entero = int(numeros)
    letra_correcta = palabra[numero_entero % 23]

    if letra == letra_correcta:
        return "DNI válido"
    else:
        return "DNI inválido"

dni = input("Introduce tu DNI: ")
resultado = validar_dni(dni)
print(resultado)
`````
