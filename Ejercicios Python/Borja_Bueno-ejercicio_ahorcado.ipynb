## Juego del Ahorcado
El funcionamiento es que vaya mostrando por pantalla ```_ _ _ _ _``` las rayas bajas correspondientes al número de letras que tiene la palabra a adivinar. El usuario deberá insertar una letra y, si esa letra existe, se mostrará o mostrarán las letras acertadas ```S _ S _ _```. 

Si falla, se mostrará el número de errores que lleva y, si lleva 6 errores, entonces se mostrará "Fin del juego".
Encapsúlalo en una función.

````python
errores = 0
aciertos = 0

palabra = input("Introduzca una palabra: ").upper()
rayas = ["_"] * len(palabra)
print(" ".join(rayas))

while errores < 6:
    adivinar_letra = input("Introduce una letra:").upper()

    if adivinar_letra not in palabra:
        errores += 1
        if errores < 5:
            print(f"Has fallado. Tienes {6 - errores} intentos restantes")
        else:
            print("Te han ahorcado")
    else:
        aciertos += 1
        for i, letra in enumerate(palabra):
            if letra == adivinar_letra:
                rayas[i] = adivinar_letra

    print(" ".join(rayas))
    if '_' not in rayas:
        print("Has ganado")
        errores += 6
````
