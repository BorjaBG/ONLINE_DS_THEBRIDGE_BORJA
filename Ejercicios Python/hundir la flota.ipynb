```python
class Barco:
    def __init__(self, id: int, tipo: str, longitud: int, x: int, y: int, orientacion: str, vidas: int):
        self.id = id
        self.tipo = tipo
        self.longitud = longitud
        self.x = x
        self.y = y
        self.orientacion = orientacion
        self.vidas = vidas

class Jugador:
    def __init__(self, nombre: str, barcos: list[Barco], es_maquina: bool = False):
        self.nombre = nombre
        self.barcos = barcos
        self.es_maquina = es_maquina
        self.tablero = Tablero.crear_tablero()
        self.mapa_barcos = {}
        self.colocar_barcos()

    def colocar_barcos(self):
        for barco in self.barcos:
            cont = 0
            x = barco.x
            y = barco.y
            longitud = barco.longitud
            while cont < longitud:
                if self.es_maquina:
                    self.mapa_barcos[(x, y)] = barco
                    if barco.orientacion == 'H':
                        y +=1
                    else:
                        x +=1
                    cont += 1
                else:
                    self.mapa_barcos[(x, y)] = barco
                    if longitud == 1:
                        self.tablero[x][y] = "B"
                    elif barco.orientacion == 'H':
                        self.tablero[x][y] = "B"
                        y +=1
                    else:
                        self.tablero[x][y] = "B"
                        x +=1
                    cont += 1

class Tablero:
    def __init__(self, id: int, dimension: int, barcos):
            self.id_jugador = id
            self.dimension = dimension
            self.barcos = barcos

    @staticmethod
    def crear_tablero():
        tablero = []
        for i in range(10):
            linea = []
            for j in range(10):
                linea.append("A")
            tablero.append(linea)
        return tablero

    @staticmethod
    def disparar(tablero, mapa_barcos):
        
        bucle = True
        while bucle:
            disparo = True
            coor_x = int(input("Introduzca la coordenada X a la que disparar:"))
            coor_y = int(input("Introduzca la coordenada Y a la que disparar:"))
            
            if coor_x > 9 or coor_x < 0:
                print("La coordenada X no puede ser menor a 0 ni mayor a 9")
                disparo = False
            elif coor_y > 9 or coor_y < 0:
                print("La coordenada Y no puede ser menor a 0 ni mayor a 9")
                disparo = False
            elif tablero[coor_x][coor_y] in ['D', 'X']:
                disparo = False
                print("Ya has disparado anteriormente a esa coordenada")

            if disparo:
                vidas = 0
                bucle = False
                if (coor_x, coor_y) in mapa_barcos:
                    barco = mapa_barcos[(coor_x, coor_y)]
                    barco.vidas -= 1
                    vidas = barco.vidas
                    tablero[coor_x][coor_y] = "X"
                else:
                    vidas = -1
                    tablero[coor_x][coor_y] = "D"
                print(f"X: {coor_x}\tY: {coor_y}")
                return tablero, vidas
```
