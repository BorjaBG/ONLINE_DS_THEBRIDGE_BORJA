## Ejercicio 2
Escribe un programa para pasar de grados a radianes. Hay que usar `input`. Recuerda que la conversión se realiza mediante

$$radianes = grados*(pi/180)$$

````python
import math
````

````python
GRADOS = "Introduce los grados: "
grados = float(input(GRADOS))

radianes = grados * (math.pi / 180)
print(f"{grados} grados son {radianes} radianes")
````

## Ejercicio 6
Mismo programa que en 5, pero en esta ocasión tienen que ser tres inputs y dos salidas. Una de las salidas que nos indique si todos son iguales, y la otra si al menos dos inputs sí que lo son

````python
INP_3 = "Introduce algo más: "

input_1 = input(INP_1)
input_2 = input(INP_2)
input_3 = input(INP_3)

if input_1 == input_2 and input_1 == input_3:
    print("Los tres son iguales")
else:
    print("Ninguno de los tres es igual a los otros dos")

if input_1 == input_2 or input_1 == input_3 or input_2 == input_3:
    print("Al menos dos son iguales")
else:
    print("No hay ningun input igual")
````

## Ejercicio 8
Razona sin ejecutar código el output que obtendremos de las siguientes sentencias
1. True and True and False
2. not ((True or False) and (True or False))
3. (False or False or False or True) and False
4. not False and True and not True

Compruébalo después con python

1. Es False porque para ser True, las tres deberían serlo
2. Es False porque aunque dentro del parentesis salga True, por fuera le estamos aplicando *not*, que es False
3. Lo que esta entre parentesis queda en True, pero con el False de fuera, al final el resultado es False
4. Aunque la primera y segunda condicion sean True, la ultima sigue siendo False y por eso el resultado final es False

````python
print(True and True and False)
print(not((True or False) and (True or False)))
print((False or False or False or True) and False)
print(not False and True and not True)
````

## Ejercicio 10
Para el siguiente string se pide imprimir por pantalla los siguientes casos

    "Lo primero de todo, cómo están los máquinas"

1. Pasarlo todo a mayúsculas
2. Pasarlo todo a minúsculas
3. Pasar a mayúsculas solo las iniciales de las palabras
4. Crea una lista dividiendolo por sus espacios
5. Sustituye las comas `,` por puntos y comas `;`
6. Elimina las `a` minúsculas

````python
texto = "Lo primero de todo, cómo están los máquinas"

print(texto.upper())
print(texto.lower())
print(texto.capitalize())

lista = texto.split()
print(lista)

print(texto.replace(",", ";"))
print(texto.replace("a", ""))
````
