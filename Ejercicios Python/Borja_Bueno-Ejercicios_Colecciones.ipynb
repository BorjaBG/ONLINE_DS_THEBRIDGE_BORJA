## Ejercicio 1
Dada la siguiente lista:
> ```ejer_1 = [1,2,3,4,5]```

Inviértela para que quede de la siguiente manera
> ```ejer_1 = [5,4,3,2,1]```

````python
ejer_1 = [1,2,3,4,5]
ejer1_1 = ejer_1.copy()
ejer1_1.reverse()


print(f"Lista revertida:\n{ejer1_1}")
````

Otra forma

````python
ejer1_1 = ejer_1.copy()
ejer1_1.sort(reverse=True)
print(f"Lista revertida:\n{ejer1_1}")
````

Otra forma

````python
ejer1_1 = ejer_1.copy()
ejer1_1 = ejer1_1[::-1]
print(f"Lista revertida:\n{ejer1_1}")
````

## Ejercicio 2
Eleva todos los elementos de la lista al cuadrado
> ```ejer_2 = [1,2,3,4,5]```

````python
ejer_2 = [1,2,3,4,5]
ejer_2_2 = []

for num in ejer_2:
    cuadrado = num **2
    ejer_2_2.append(cuadrado)

print(f"Esta es la lista de los numeros elevados al cuadrado: \n{ejer_2_2}")
````

## Ejercicio 3
Crea una lista nueva con todas las combinaciones de las siguientes dos listas:
> ```ejer_3_1 = ["Hola", "amigo"]```
>
> ```ejer_3_2 = ["Que", "tal"]```

Obten el siguiente output:

```['Hola Que', 'Hola tal', 'amigo Que', 'amigo tal']```

````python
ejer_3_1 = ["Hola", "amigo"]
ejer_3_2 = ["Que", "tal"]
lista_palabras = []

for palabra in ejer_3_1:
    for palabra_2 in ejer_3_2:
        lista_palabras.append(palabra + " " + palabra_2)

print(lista_palabras)
````

## Ejercicio 4
Dada la siguiente lista, encuentra por el índice el valor 45 y sustitúyelo por el 0
> ```ejer_4 = [20, 47, 19, 29, 45, 67, 78, 90]```

## Ejercicio 5
Dada la siguiente lista, elimina todos los valores iguales a 3
> ```ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]```

TIP: No intentes eliminar elementos sobre la lista que estás iterando. Haz una copia con `ejer_5.copy()`.

````python
ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]
ejer_5_5 = ejer_5.copy()
bucle = True

while bucle:
    if (3 in ejer_5_5):
        ejer_5_5.remove(3)
    else:
        bucle = False

print(ejer_5_5)
````

## Ejercicio 11
Convierte la siguiente lista en un diccionario
> ```ejer_11 = [("x", 1), ("x", 2), ("x", 3), ("y", 1), ("y", 2), ("z", 1)]```

TIP: Puedes usar dos variables de control en un bucle `for`. Una por cada valor de las tuplas.

````python
ejer_11 = [("x", 1), ("x", 2), ("x", 3), ("y", 1), ("y", 2), ("z", 1)]
diccionario_ejer_11 = {}

for letra, numero in ejer_11:
    if letra not in diccionario_ejer_11:
        diccionario_ejer_11[letra] = []
    diccionario_ejer_11[letra].append(numero)

print(diccionario_ejer_11)
````

## Ejercicio 16
