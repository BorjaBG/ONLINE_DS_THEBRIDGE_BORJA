```python
MSG_MENU = "¡Bienvenido a la biblioteca! Elige que quieres hacer (1-6):\n 1- Listar todos los libros\n 2- Buscar libro\n 3- Añadir libro\n 4- Eliminar libro por titulo\n 5- Alquilar libro por titulo\n 6- Devolver libro por titulo"
MSG_CABECERA = f"{'Titulo':<75}{'Autor':<70}{'Alquilado':<10}\n{'-'*160}\n"
MSG_BUSQUEDA = "¿Que libro desea buscar? Introduzca el titulo del libro: "
MSG_TITULO_NUEVO = "Introduzca el titulo del libro que desea añadir: "
MSG_AUTOR_NUEVO = "Introduzca el autor del libro que desea añadir: "
MSG_LIBRO_ANADIDO = "Acaba de añadir el siguiente libro a la lista:\n"
MSG_TITULO_ELIMINAR = "Introduzca el titulo del libro que desea eliminar: "
MSG_LIBRO_DUPLICADO = "No se ha encontrado el libro o hay dos o más libros con el mismo nombre."
MSG_LIBRO_NO_ENCONTRADO = "\nNo hay libros que cumplan este criterio.\n"

MSG_ERROR_MENU = "La opcion que ha elegido no es valida, elija una opcion del menu del 1 al 6"

SIGNOS_PUNTUACION = ",-.:;!?'\"()[]{}"
```

```python


def listar_libros(lista, filtro_alquilado = None):
    menu = MSG_CABECERA
    libros_mostrados = 0

    for libro in lista:
        if filtro_alquilado is not None and libro.get("Alquilado") != filtro_alquilado:
            continue

        libros_mostrados += 1
        if libro.get("Alquilado"):
            alquilado = "Sí"
        else:
            alquilado = "No"
        menu = menu + f"{libro.get("Titulo"):<75}{libro.get("Autor"):<70}{alquilado:<10}\n"

    if libros_mostrados == 0:
        print(MSG_LIBRO_NO_ENCONTRADO)
    else:
        print(menu)

def buscar_libros(lista):
    resultados = []
    busqueda = input(MSG_BUSQUEDA).strip().lower()
    for libro in lista:
        titulo = libro.get("Titulo").lower()
        titulo_limpio = "".join(char for char in titulo if char not in SIGNOS_PUNTUACION)
        busqueda_limpia = "".join(char for char in busqueda if char not in SIGNOS_PUNTUACION)

        if busqueda_limpia in titulo_limpio:
            resultados.append(libro)

    listar_libros(resultados)

def gestionar_libros(lista, accion: str):
    completado  = False

    if accion == "alquilar":
        listar_libros(lista, filtro_alquilado=False)
    elif accion == "devolver":
        listar_libros(lista, filtro_alquilado=True)

    while not completado:
        duplicado = False
        cont = 0
        indice_libro = 0

        titulo_usuario = input(f"Introduzca el titulo del libro que desea {accion}: ")
        titulo_usuario_copy = titulo_usuario.strip().lower()

        for indice, libro in enumerate(lista):
            titulo = libro.get("Titulo").lower()
            titulo_limpio = "".join(char for char in titulo if char not in SIGNOS_PUNTUACION)
            titulo_usuario_copy = "".join(char for char in titulo_usuario_copy if char not in SIGNOS_PUNTUACION)
            
            if titulo_usuario_copy in titulo_limpio:
                titulo_usuario = libro.get("Titulo")
                indice_libro = indice
                cont += 1
                if cont > 1:
                    duplicado = True

        if cont == 0:
            duplicado = True

        if duplicado:
            print(MSG_LIBRO_DUPLICADO)
        else:
            match accion:
                case "eliminar":
                    lista.pop(indice_libro)
                    print(f"El libro {titulo_usuario} ha sido eliminado.")
                case "alquilar":
                    lista[indice_libro]["Alquilado"] = True
                    print(f"El libro {titulo_usuario} ha sido alquilado.")
                case "devolver":
                    lista[indice_libro]["Alquilado"] = False
                    print(f"El libro {titulo_usuario} ha sido devuelto.")
            completado = True

try:
    eleccion = int(input(MSG_MENU))

    match eleccion:
        case 1:
            listar_libros(libros)
        case 2:
            buscar_libros(libros)
        case 3:
            titulo_nuevo = input(MSG_TITULO_NUEVO)
            autor_nuevo = input(MSG_AUTOR_NUEVO)
            libros.append({"Titulo": titulo_nuevo, "Autor": autor_nuevo, "Alquilado": False})
            print(f"{MSG_LIBRO_ANADIDO}Titulo: {titulo_nuevo}\tAutor: {autor_nuevo}")
        case 4:
            gestionar_libros(libros, "eliminar")
        case 5:
            gestionar_libros(libros, "alquilar")
        case 6:
            gestionar_libros(libros, "devolver")
        case _:
            print(MSG_ERROR_MENU)

except Exception as e:
    print("¡Error!")
```

```python
MSG_MENU = "\n¡Bienvenido a la biblioteca! Elige que quieres hacer (0-6):\n 1- Listar todos los libros\n 2- Buscar libro\n 3- Añadir libro\n 4- Eliminar libro por ID\n 5- Alquilar libro por ID\n 6- Devolver libro por ID\n 0- Salir de la biblioteca\nOpción: "
MSG_CABECERA = f"{'ID':<6}{'Titulo':<75}{'Autor':<70}{'Alquilado':<10}\n{'-'*166}\n"
MSG_BUSQUEDA = "Que libro desea buscar? Introduzca el titulo del libro: "
MSG_TITULO_NUEVO = "Introduzca el titulo del libro que desea añadir: "
MSG_AUTOR_NUEVO = "Introduzca el autor del libro que desea añadir: "
MSG_LIBRO_ANADIDO = "Acaba de añadir el siguiente libro a la lista:\n"
MSG_LIBRO_DUPLICADO = "No se ha encontrado ningún libro con el ID indicado."
MSG_LIBRO_NO_ENCONTRADO = "\nNo hay libros que cumplan este criterio.\n"
MSG_ERROR_MENU = "La opcion que ha elegido no es valida, elija una opcion del menu del 0 al 6"
SIGNOS_PUNTUACION = ",-.:;!?'\"()[]{}"

class Libro:
    def __init__(self, id_libro: int, titulo: str, autor: str, alquilado: bool = False):
        self.id_libro = id_libro
        self.titulo = titulo
        self.autor = autor
        self.alquilado = alquilado

    def alquilar(self):
        if not self.alquilado:
            self.alquilado = True
            return True
        return False

    def devolver(self):
        if self.alquilado:
            self.alquilado = False
            return True
        return False

    def titulo_limpio(self):
        return "".join(char for char in self.titulo.lower() if char not in SIGNOS_PUNTUACION)

    def coincide_con(self, busqueda):
        busqueda_limpia = "".join(char for char in busqueda.lower() if char not in SIGNOS_PUNTUACION)
        return busqueda_limpia in self.titulo_limpio()

    def __str__(self):
        estado = "Sí" if self.alquilado else "No"
        return f"{self.id_libro:<6}{self.titulo:<75}{self.autor:<70}{estado:<10}\n"


class Biblioteca:
    def __init__(self, libros):
        self.libros = libros
        self.ultimo_id = max((l.id_libro for l in self.libros), default=0)

    def anadir_libro(self, titulo: str, autor: str):
        self.ultimo_id += 1
        nuevo = Libro(self.ultimo_id, titulo, autor, alquilado=False)
        self.libros.append(nuevo)
        return nuevo

    def obtener_libro_por_id(self, id_buscar: int):
        for indice, libro in enumerate(self.libros):
            if libro.id_libro == id_buscar:
                return indice, libro
        return None, None

    def listar_libros(self, libros_a_mostrar=None, filtro_alquilado=None):
        lista = libros_a_mostrar if libros_a_mostrar is not None else self.libros
        menu = MSG_CABECERA
        libros_mostrados = 0

        for libro in lista:
            if filtro_alquilado is not None and libro.alquilado != filtro_alquilado:
                continue
            libros_mostrados += 1
            menu += str(libro)

        if libros_mostrados == 0:
            return MSG_LIBRO_NO_ENCONTRADO
        return menu

    def buscar_libros(self):
        resultados = []
        busqueda = input(MSG_BUSQUEDA).strip().lower()
        for libro in self.libros:
            if libro.coincide_con(busqueda):
                resultados.append(libro)
        print(self.listar_libros(resultados))

    def gestionar_libros(self, accion: str):
        completado = False

        if accion == "alquilar":
            print("\nLibros disponibles:")
            print(self.listar_libros(filtro_alquilado=False))
        elif accion == "devolver":
            print("\nLibros alquilados:")
            print(self.listar_libros(filtro_alquilado=True))

        while not completado:
            entrada_usuario = input(f"Introduzca el ID del libro que desea {accion}: ").strip()
            
            if not entrada_usuario.isdigit():
                print(MSG_LIBRO_DUPLICADO)
                continue

            id_libro = int(entrada_usuario)
            indice_libro, libro = self.obtener_libro_por_id(id_libro)

            if libro is None:
                print(MSG_LIBRO_DUPLICADO)
            else:
                match accion:
                    case "eliminar":
                        print(f"El libro '{libro.titulo}' (ID {libro.id_libro}) ha sido eliminado.")
                        self.libros.pop(indice_libro)
                    case "alquilar":
                        if libro.alquilar():
                            print(f"El libro '{libro.titulo}' (ID {libro.id_libro}) ha sido alquilado.")
                        else:
                            print(f"El libro '{libro.titulo}' (ID {libro.id_libro}) ya estaba alquilado.")
                    case "devolver":
                        if libro.devolver():
                            print(f"El libro '{libro.titulo}' (ID {libro.id_libro}) ha sido devuelto.")
                        else:
                            print(f"El libro '{libro.titulo}' (ID {libro.id_libro}) ya estaba disponible.")
                completado = True

listado_libros = [
    Libro(indice + 1, libro["Titulo"], libro["Autor"], libro["Alquilado"]) 
    for indice, libro in enumerate(libros)
]
biblioteca = Biblioteca(listado_libros)

while True:
    try:
        entrada = input(MSG_MENU).strip()
        if not entrada.isdigit():
            print(MSG_ERROR_MENU)
            continue

        eleccion = int(entrada)

        match eleccion:
            case 0:
                print("\n¡Gracias por utilizar la biblioteca! Hasta pronto.\n")
                break
            case 1:
                print(biblioteca.listar_libros())
            case 2:
                biblioteca.buscar_libros()
            case 3:
                titulo_nuevo = input(MSG_TITULO_NUEVO)
                autor_nuevo = input(MSG_AUTOR_NUEVO)
                nuevo_libro = biblioteca.anadir_libro(titulo_nuevo, autor_nuevo)
                print(f"{MSG_LIBRO_ANADIDO}ID: {nuevo_libro.id_libro}\tTitulo: {nuevo_libro.titulo}\tAutor: {nuevo_libro.autor}")
            case 4:
                biblioteca.gestionar_libros("eliminar")
            case 5:
                biblioteca.gestionar_libros("alquilar")
            case 6:
                biblioteca.gestionar_libros("devolver")
            case _:
                print(MSG_ERROR_MENU)
    except Exception as e:
        print("¡Error!", e)
```
