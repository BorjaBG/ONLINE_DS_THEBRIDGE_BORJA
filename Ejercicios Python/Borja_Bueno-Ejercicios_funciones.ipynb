## Ejercicio 1
Escribe una función que convierta números del 1 al 7 en nombres de los dias de la semana. La función constará de un único argumento numérico y una salida de tipo string

````python
def convertir_numeros(num: int):
    '''
        Funcion para convertir un numero en el dia de la semana.
        Args:
            num (int): Numero del dia de la semana
        Return:
            str: Dia de la semana
    '''
    match num:
        case 1:
            return "Lunes"
        case 2:
            return "Martes"
        case 3:
            return "Miercoles"
        case 4:
            return "Jueves"
        case 5:
            return "Viernes"
        case 6:
            return "Sabado"
        case 7:
            return "Domingo"
        case _:
            return "¡Error! Ese numero de la semana no es válido"

dia = int(input("Introduzca el dia de la semana (1-7):"))

print(convertir_numeros(dia))
````

## Ejercicio 2
En el ejercicio 8 de flujos de control, creábamos una pirámide invertida de números desde el 5. Crea una función que replique el comportamiento de la pirámide, y utiliza un único parámetro de entrada de la función para determinar el número de filas de la pirámide.

````python
lista = [8, 7, 6, 5, 4, 3, 2, 1]

def piramide(lista_piramide: list):
    lista_impresa = ""
    imprimir = True

    while imprimir:
        for num in lista_piramide:
            lista_impresa = lista_impresa + f" {str(num)}"
        print(f"{lista_impresa}\n")
        lista_piramide.pop(0)
        lista_impresa = ""

        if len(lista_piramide) == 0:
            imprimir = False

piramide(lista)
````

## Ejercicio 3
Escribe una función que compare dos números. La función tiene dos argumentos y hay tres salidas posibles: que sean iguales, que el primero sea  mayor que el segundo, o que el segundo sea mayor que el primero

````python
def comparar_numeros(num1: float, num2: float):
    if num1 > num2:
        print(f"{num1} es mayor que {num2}")
    elif num2 > num1:
        print(f"{num2} es mayor que {num1}")
    else:
        print(f"{num1} y {num2} son iguales")

try:
    numero1 = float(input("Introduzca un numero:"))
    numero2 = float(input("Introduzca otro numero:"))

    comparar_numeros(numero1, numero2)
except ValueError:
    print("¡Error! Eso no es un numero.")
````

## Ejercicio 4
Escribe una función que sea un contador de letras. En el primer argumento tienes que introducir un texto, y el segundo que sea una letra que contar. La función tiene que devolver un entero con el número de veces que aparece esa letra, tanto mayúscula, como minúscula

````python
def contar_letras(texto: str, letra_buscar: str):
    letra_mayuscula = 0
    letra_minuscula = 0
    for letra in texto:
        if letra == letra_buscar.lower():
            letra_minuscula += 1
        elif letra == letra_buscar.upper():
            letra_mayuscula += 1
    resultado = letra_mayuscula + letra_minuscula
    return resultado

SIGNOS_PUNTUACION = ",-.:;!?'\"()[]{}"
texto = "Lorem ipsum dolor sit amet, consectetur adipiscing elit. Fusce suscipit sed massa ut tempor. Aliquam bibendum elementum viverra. Sed dolor."
texto_limpio = texto.strip()
texto_limpio = "".join(char for char in texto_limpio if char not in SIGNOS_PUNTUACION)

print(texto)
letra = input("Introduzca la letra que quiere buscar en el texto:")
total = contar_letras(texto_limpio, letra)
print(f"La letra {letra} se encuentra {total} veces en el texto")
````

## Ejercicio 5
Escribe una función que tenga un único argumento, un string. La salida de la función tiene que ser un diccionario con el conteo de todas las letras de ese string, siendo la clave la letra y el valor el conteo.

````python
def contar(frase: str):
    dic_letras = {}

    for letra in frase:
        if letra not in dic_letras:
            dic_letras[letra] = 1
        else:
            dic_letras[letra] = dic_letras[letra] + 1

    return dic_letras

SIGNOS_PUNTUACION = ",-.:;!?'\"()[]{}"
texto = "Lorem ipsum dolor sit amet, consectetur adipiscing elit. Sed nulla mauris, pellentesque quis diam vel, aliquet egestas ipsum. Etiam fermentum fringilla facilisis. Sed aliquet ligula accumsan, accumsan urna porta, accumsan ligula. Mauris a volutpat felis. Nam convallis, libero a ornare lacinia, est nulla porttitor nunc."
texto_limpio = texto.strip().replace(" ", "")
texto_limpio = "".join(char for char in texto_limpio if char not in SIGNOS_PUNTUACION)
diccionario = contar(texto_limpio)
print(diccionario)
````

## Ejercicio 6
Escribir una función que añada o elimine elementos en una lista. La función necesita los siguientes argumentos:
* lista: la lista donde se añadirán o eliminarán los elementos
* comando: "add" o "remove"
* elemento: Por defecto es None.

Tendrá como salida la lista

````python
def modificar_elemento(lista: list, comando: str, elemento = None):
    match comando.lower():
        case "add":
            lista.append(elemento)
            return lista
        case "remove":
            lista.remove(elemento)
            return lista

lista = ['patata', 'miercoles', "6", "24", 'ventana', "3", "7", "1"]

print(lista)
accion = input("¿Desea añadir o eliminar un elemento de la lista? (add o remove)")

if accion == "add":
    palabra = input("Introduzca lo que quiera para añadir a la lista:")
    lista_mod = modificar_elemento(lista, accion, palabra)
elif accion == "remove":
    palabra = input("Introduzca lo que quiera eliminar a la lista:")
    lista_mod = modificar_elemento(lista, accion, palabra)
else:
    print("Eso no es un comando valido. Solo son validos add y remove")

print(lista_mod)
````

## Ejercicio 7
Crea una función que reciba un número arbitrario de palabras, y devuelva una frase completa, separando las palabras con espacios.

````python
def crear_frase(lista: list):
    frase = ""

    for palabra in lista:
        frase = frase + " " + palabra
    return frase

continuar = True
lista_palabras = []

while continuar:
    palabra = input("Introduzca una palabra o introduzca 0 si quiere salir:")

    if palabra != "0":
        lista_palabras.append(palabra)
    else:
        continuar = False

frase = crear_frase(lista_palabras)
print(frase)
````

## Ejercicio 8
Escribe un programa que obtenga el enésimo número de la [serie de Fibonacci](https://es.wikipedia.org/wiki/Sucesi%C3%B3n_de_Fibonacci). Tienes que crear una función recursiva con un único argumento.

```Python
fibonacci(2) = 1
fibonacci(4) = 3
fibonacci(5) = 5
fibonacci(8) = 21
```

fibonacci(n) = fibonacci(n-1) + fibonacci(n-2)

````python
def fibonacci(num: int):
    if num <= 0:
        return 0
    elif num == 1:
        return 1
    elif num == 2:
        return 1
    else:
        return fibonacci(num - 1) + fibonacci(num - 2)

numero = int(input("Introduzca un numero: "))
print(fibonacci(numero))
````

## Ejercicio 9
Define en una única celda las siguientes funciones:
* Función que calcule el área de un cuadrado
* Función que calcule el área de un triángulo
* Función que calcule el área de un círculo

En otra celda, calcular el area de:
* Dos círculos de radio 10 + un triángulo de base 3 y altura 7
* Un cuadrado de lado = 10 + 3 círculos (uno de radio = 4 y los otros dos de radio = 6) + 5 triángulos de base = 2 + altura = 4

````python
import math
````

````python
def area_cuadrado(lado: float):
    area = lado**2
    return area

def area_triangulo(base: float, altura: float):
    area = (base * altura) / 2
    return area

def area_circulo(radio: float):
    area = math.pi * (radio**2)
    return area
````

````python
print("Dos círculos de radio 10 + un triángulo de base 3 y altura 7")
area_circulo_1 = area_circulo(10)
area_circulo_2 = area_circulo(10)
area_triangulo_1 = area_triangulo(3, 7)
area_total = area_circulo_1 + area_circulo_2 + area_triangulo_1
print(f"El area total es de {area_total}")

print("Un cuadrado de lado = 10 + 3 círculos (uno de radio = 4 y los otros dos de radio = 6) + 5 triángulos de base = 2 + altura = 4")
area_cuadrado_1 = area_cuadrado(10)
area_circulo_3 = area_circulo(4)
area_circulo_4 = area_circulo(6)
area_circulo_5 = area_circulo(6)
area_triangulo_2 = 0

for i in range(0, 5):
    area_triangulo_2 = area_triangulo_2 + area_triangulo(2, 4)

area_total_2 = area_cuadrado_1 + area_circulo_3 + area_circulo_4 + area_circulo_5 + area_triangulo_2
print(f"El area total es de {area_total_2}")
````
