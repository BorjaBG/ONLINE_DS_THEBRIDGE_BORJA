## Ejercicio Extra
Programa una calculadora que

* Pregunte qué operación quieres realizar (multiplicación/división/suma/resta/potencia/salir)
* Pida dos números 
* Realice la operación
* Siga funcionando hasta que el usuario elija salir

#### Bonus:
* Que guarde todas las operaciones realizadas y las imprima por pantalla cuando el usuario elija esa opción
* Que sea robusta frente a entradas erróneas del usuario o frente a operaciones como dividir por 0

```python
import math
import pprint
import time
```

```python
MENU = "¿Que operación quiere realizar? \n1- Sumar \n2- Restar \n3- Multiplicación \n4- División \n5- Potencia \n6- Salir \n7- Ver historial"
TXT_1_NUMERO = "Introduzca el primer numero de la operación: "
TXT_2_NUMERO = "Introduzca el segundo numero de la operación: "
TXT_RESULTADO = "El resultado de la %s de los numeros %s y %s es %s"
MSG_LOG = "Estas son las operaciones que has realizado: "
MSG_DESPEDIDA = "Gracias por usar esta calculadora ¡Agur!"
MSG_ERROR = "¡Error! Introduzca una de las opciones numericas del menu"
MSG_ERROR_MENU = "Esa no es una eleccion valida"

eleccion = 1
salir = False
log = []
log_resultado = ""

def realizar_operaciones(op_elegida):
    '''
        Funcion que realiza la operación matemática elegida por el usuario.
        Args:
            op_elegida (int): Número que representa la operación elegida por el usuario.
        Returns:
            None
    '''
    prim_numero = float(input(TXT_1_NUMERO))
    seg_numero = float(input(TXT_2_NUMERO))
    tipo_op = ""
    operacion = ""

    match op_elegida:
        case 1:
            tipo_op = "suma"
            operacion = '+'
            resultado = prim_numero + seg_numero
        case 2:
            tipo_op = "resta"
            operacion = '-'
            resultado = prim_numero - seg_numero
        case 3:
            tipo_op = "multiplicacion"
            operacion = '*'
            resultado = prim_numero * seg_numero
        case 4:
            if(seg_numero != 0):
                tipo_op = "division"
                operacion = '/'
                resultado = prim_numero / seg_numero
            else:
                print("No es posible dividir entre 0")
                log_resultado = "No es posible dividir entre 0"
        case 5:
            tipo_op = "potencia"
            operacion = '^'
            resultado = prim_numero ** seg_numero

    print(TXT_RESULTADO % (tipo_op, prim_numero, seg_numero, resultado))
    log_resultado = f"{prim_numero} {operacion} {seg_numero} = {resultado}"
    log.append(log_resultado)
    return

try:
    while eleccion > 0 and eleccion < 6 and not salir:
        eleccion = float(input(MENU))        

        match eleccion:
            case 1:
                realizar_operaciones(1)
            case 2:
                realizar_operaciones(2)
            case 3:
                realizar_operaciones(3)
            case 4:
                realizar_operaciones(4)
            case 5:
                realizar_operaciones(5)
            case 6:
                print(MSG_DESPEDIDA)
            case 7:
                print(MSG_LOG)
                pprint.pprint(log)
            case _:
                print(MSG_ERROR_MENU)
                eleccion = 1
                continue
except ValueError:
    print(MSG_ERROR)
```
